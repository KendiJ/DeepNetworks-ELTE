### Exercise 2.1: Rectangle and Square classes

#### Problem Description
In object-oriented programming, subclasses can override parent methods to enforce specific domain constraints or provide a more convenient interface. However, changing a method's signature requires care so that code expecting the parent class does not break.

You are tasked with working with a base `Rectangle` class and creating a derived `Square` class:

1. **Base Class (`Rectangle`)**:
   - Constructor (`__init__`) accepts `width` and `height`.
   - Method `area()` returns the area of the rectangle.
   - Method `perimeter()` returns the perimeter of the rectangle.
   - Method `resize(new_width, new_height)` updates dimensions.
   - Static method `is_valid_dimension(val)` ensures that `val` is a positive number (and thus would be a valid dimension for a rectangle).
   - Dunder methods `__str__` (for printing) and `__gt__` (comparing areas).

2. **Derived Class (`Square`)**:
   - Inherits from `Rectangle`.
   - Constructor (`__init__`) accepts a single `side_length` and calls `super().__init__(side_length, side_length)`.
   - Override `resize` behavior while maintaining the function signature: raise a `ValueError` if two unequal dimensions are passed to a `Square.resize()`.

````python
# Implement your solution in this cell
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)
    
    def resize(self, new_width, new_height):
        self.width = new_width
        self.height = new_height

    @staticmethod
    def is_valid_dimension(val):
        return isinstance(val, (int, float)) and val > 0

    def __str__(self):
        return f"Rectangle(width={self.width}, height={self.height})"

    def __gt__(self, other):
        return self.area() > other.area()

class Square(Rectangle):
    def __init__(self, side_length):
        super().__init__(side_length, side_length)

    def resize(self, new_width, new_height):
        if new_width != new_height:
            raise ValueError("square should have equal dimentions")
        super().resize(new_width, new_height)




````

### Exercise 2.2: Integer sliding window iterator

#### Problem Description
Implement the `SlidingWindowIterator` class which is both an iterable and an iterator. It needs to yield lists of consecutive integers from `k` to `k+window_len-1` (inclusive), where `k` goes from 0 to `n-window_len` (inclusive).

Implement the following functionalities:
- Constructor (`__init__`) accepts `n` and `window_len`.
- `__iter__` will return self.
- `__next__` will return the next integer list in order containing `window_len` subsequent integers. In the first iteration the list starts from 0. In each following iteration the starting element is increased by 1. The iteration ends when the last element of the returned list is equal to `n-1`. If `n` is smaller than `sindow_len`, then no lists are returned.

#### Sample output
```
>> it = SlidingWindowIterator(5,3)
>> for item in it:
>>   print("    ", item)

[0, 1, 2]
[1, 2, 3]
[2, 3, 4]
```

````python
# Implement your solution in this cell
class SlidingWindowIterator:
    def __init__(self, n, window_len):
        self.n  = n
        self.window_len = window_len
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current + self.window_len > self.n:
            raise StopIteration
        window = list(range(self.current, self.current + self.window_len))
        self.current += 1
        return window





````

#### Test Cases for Exercise 2.2

````python
# Tests for Exercise 2.2

it1 = SlidingWindowIterator(5, 3)
assert iter(it1) is it1, "__iter__ must return self"
assert next(it1) == [0, 1, 2], "First window should be [0, 1, 2]"
assert next(it1) == [1, 2, 3], "Second window should be [1, 2, 3]"
assert next(it1) == [2, 3, 4], "Third window should be [2, 3, 4]"

stop_raised = False
try:
    next(it1)
except StopIteration:
    stop_raised = True
assert stop_raised is True, "Iterator must raise StopIteration when finished"

it2 = SlidingWindowIterator(5, 3)
result2 = list(it2)
expected2 = [[0, 1, 2], [1, 2, 3], [2, 3, 4]]
assert result2 == expected2, f"Expected {expected2}, but got {result2}"

it3 = SlidingWindowIterator(4, 4)
result3 = list(it3)
expected3 = [[0, 1, 2, 3]]
assert result3 == expected3, f"Expected {expected3}, but got {result3}"

it4 = SlidingWindowIterator(3, 1)
result4 = list(it4)
expected4 = [[0], [1], [2]]
assert result4 == expected4, f"Expected {expected4}, but got {result4}"

it5 = SlidingWindowIterator(3, 5)
result5 = list(it5)
expected5 = []
assert result5 == expected5, f"Expected empty list when window_len > n, got {result5}"

result_spent = list(it1)
assert result_spent == [], "Exhausted iterator should yield no further elements"

print("Exercise 2.2: All tests passed!")
````
