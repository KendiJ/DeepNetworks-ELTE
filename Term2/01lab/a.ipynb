```python
def c_range(start, stop, step): 
    for i in range(start, stop, step):
        yield i
```

```python
gen = c_range(1, 100, 2)
```

```python
next(gen)
```

```python
list(gen)
```

```python
class Counter:
    def __init__(self, start, stop):
        self.start = start
        self.stop = stop

    def __repr__(self):
        return f'Counter(start={self.start}, {self.stop})'
```

```python
counter = Counter(1, 100)
```
