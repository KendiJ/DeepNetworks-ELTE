```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('./games_genres.csv', index_col=0)
df.drop_duplicates(subset=['Title'],inplace=True)
df.head()
```

# Which game received the highest rating?

- np.max
- element-wise comparison

```python
max_rate = np.max(df["Rating"])
highest_rated_games = df[df["Rating"] == max_rate]

print(highest_rated_games, f'this was the highest')
```

- np.argmax

# What is the average rating, rounded to two decimal places?

- np.sum
- np.ndarray.size

```python
rates_array = df["Rating"].to_numpy()

total_sum = np.sum(rates_array)
total_elements = rates_array.size

average_rating = round(total_sum / total_elements, 2)

print(f"Average Rating: {average_rating}")
```

- np.mean

# Which game belongs to the most categories?

- np.max
- np.sum along an axis
- np.argmax

```python
# game_cat = df.loc[:, "Adventure":]


game_cat = df.loc[:, "Adventure":]

df['Total_Categories'] = game_cat.sum(axis=1)

max_categories = df['Total_Categories'].max()

top_games = df[df['Total_Categories'] == max_categories]

print(f"The highest number of categories a game belongs to is: {max_categories}\n")
print("Game(s) with the most categories:")
print(top_games[['Title', 'Total_Categories']].to_string(index=False))
```

```python
category_counts = np.sum(game_cat.to_numpy(), axis=1)
max_categories = np.max(category_counts)

highest_index = np.argmax(category_counts)

top_game_title = df.loc[highest_index, "Title"]

print(f"Game with the most categories: '{top_game_title}'")
print(f"Total categories it belongs to: {max_categories}")
```
